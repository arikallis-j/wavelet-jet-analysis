```julia
using Oceananigans

grid = RectilinearGrid(topology = (Periodic, Periodic, Bounded),
                       size = (4, 5, 4),
                       halo = (1, 1, 1),
                       x = (0, 1),
                       y = (0, 1),
                       z = [0, 0.1, 0.3, 0.6, 1])
```

```julia
c = Field{Center, Center, Center}(grid)
```

```julia
c == CenterField(grid)
```

```julia
u = Field{Face, Center, Center}(grid)
```

```julia
u == XFaceField(grid)
```

```julia
w = Field{Center, Center, Face}(grid)
```

```julia
w == ZFaceField(grid)
```
