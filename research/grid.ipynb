```julia
grid = RectilinearGrid(
    size=(16,16),
    x=(0,2π),
    y=(0,2π),
    topology=(Periodic, Periodic, Flat)
)

model = NonhydrostaticModel(
    grid;
    advection = Centered(order=4),
    timestepper = :RungeKutta3,
)

simulation = Simulation(
    model,
    Δt = 0.01,
    stop_time = 1,
    progress=true,
)
```

```julia
run!(simulation)
```
